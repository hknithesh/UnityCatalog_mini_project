#### Unity Catalog
__Unity Catalog__ is a DataBricks offered *unified solution* for implementing *data governance* in the Data Lakehouse.

__Data Governance__ is the process of managing the availability, usability, integrity and security of the data present in the an enterprise.
  - controls access to the data for the users
  - ensures that the data is trust worthy and not misused
  - helps implement privacy regulations such as GDPR, CCPA etc

Unity catalog offers these Data Governance solutions
- Data Access Control
- Data Audit
- Data Lineage
- Data Discoverability

One Metastore per region is recommended by DataBricks.

##### Setup Unity Catalog
- When New Unity Catalog is created it is essential to setup new **Default Storage Account**(ADLS Gen2 Container) to save managed Tables
- Azure Connector for Databricks has to be created. This will automatically creates system managed identity and assigs managed identity to Databricks account.
- In Default Storage Account, Add Role Assignment *Storage Blob Data Contributor* Role should be assigned to Managed Identity(add members
- Link workspace(premium tier) to newly created Metastore

DataBricks Account console: https://accounts.azuredatabricks.net/

##### Glossary
- The **Azure Databricks Access Connector** lets you connect managed identities to an Azure DataBricks Account for the purpose of accessing data registered in Unity Catalog

-A **metastore** is the top-level container for catalog in Unity Catalog. Within a metastore, Unity Catalog provides a 3-level namespace for organizing data: catalogs, schemas (also called databases), and tables / views.

- Metastore admins can manage the privileges for all securable objects inside a metastore. It is recommended to assign a group as the admin when possible.

- Delta Sharing with parties outside your organization

- **Enterprise Applications** - Seamlessly integrate external applications with Azure services using pre-integrated connectors and APIs for enhanced productivity and efficiency.
