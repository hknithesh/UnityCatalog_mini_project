##### Window Function in Spark

###### Python
```
from pyspark.sql.functions import rank
from pyspark.sql.window import Window

#define a window spec with partition by and order by 
window_spec = Window.partitionBy("Name").orderBy(df["score"].desc(),df["Age"].asc())

#apply rank function
df = df.withColumn("rank", rank().over(window_spec))
```

```
SELECT Name, Score, Age, RANK() OVER(PARTITION BY NAME ORDER BY score Desc, Age) as rank FROM table 
```

Apache Airflow is an open source tool for programmatically authoring, scheduling and monitoring your data pipelines.

Benefits of using Airflow in Production:
Pipelines as code written in Python,Highly extensible, Infinitely Scalable, Large Vibrant OSS community

Featues:
Highly available scheduler
provider packages
TaskFlow API
Dynamic task mapping
Fully functional UI
Deferrable operators


Core Components:
WebServer
Scheduler
Database
Executor
Worker
Triggerer

Task Instance represents a specific run of a Task.


Operators: building blocks of airflow, operators define what that task does.

#### DBT

https://www.youtube.com/@Sleek-Insights


Tool for TRANSFORMING data within DATAWAREHOUSE.

###### run all models in the project
dbt run

###### run test for all models in the project
dbt test

###### run all models, tests, snapshots, seeds in the project
dbt build

###### execute snapshots defined in the project
dbt snapshot

###### run test for a specific model
dbt test --select my-dbt-model

###### run a specific model
dbt run --select my-dbt-model
